```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data=pd.read_csv("/content/diabetes.csv")
```

```python
data.head()
```

```python
data.isnull().sum()
```

```python
data['Pregnancies']=data['Pregnancies'].fillna(data['Pregnancies'].mean())
```

```python
data['Glucose']=data['Glucose'].fillna(data['Glucose'].mean())
```

```python
data['BloodPressure']=data['BloodPressure'].fillna(data['BloodPressure'].mean())
```

```python
data['SkinThickness']=data['SkinThickness'].fillna(data['SkinThickness'].mean())
```

```python
data['Insulin']=data['Insulin'].fillna(data['Insulin'].mean())
```

```python
data['BMI']=data['BMI'].fillna(data['BMI'].mean())
```

```python
data['DiabetesPedigreeFunction']=data['DiabetesPedigreeFunction'].fillna(data['DiabetesPedigreeFunction'].mean())
```

```python
data['Age']=data['Age'].fillna(data['Age'].mean())
```

```python
data['Outcome']=data['Outcome'].fillna(data['Outcome'].mean())
```

```python
data.isna().sum()
```

```python

from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
data['Pregnancies']=le.fit_transform(data['Pregnancies'])
data['Glucose']=le.fit_transform(data['Glucose'])
data['BloodPressure']=le.fit_transform(data['BloodPressure'])
data['SkinThickness']=le.fit_transform(data['SkinThickness'])
```

```python
data=data.drop(['DiabetesPedigreeFunction'],axis=1)
```

```python
data.head()
```

```python
x=data.drop(['Pregnancies'],axis=1) #input independent column
y=data['Pregnancies'] #output dependent column
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model=LogisticRegression()
```

```python
model.fit(x_train,y_train)
```

```python
predictions=model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score=accuracy_score(y_test,predictions)
```
